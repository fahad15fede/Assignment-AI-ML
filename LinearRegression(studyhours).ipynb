```python
import pandas as pd
```

#Importing Dataset

```python
dataset=pd.read_csv('rounded_hours_student_scores.csv')
dataset.shape
```

#Removing Outliers

```python
Q1 = dataset['Scores'].quantile(0.25)
Q3 = dataset['Scores'].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR
upper = Q3+1.5*IQR

outliers = dataset[(dataset['Scores'] < lower) | (dataset['Scores'] > upper)]
print(outliers)
dataset = dataset[(dataset['Scores'] > lower) & (dataset['Scores'] < upper)]
dataset.shape
```

#Data Preprocessing

```python
X = dataset.drop('Scores', axis =1)
y = dataset['Scores']
```

```python
# train vs test splitting
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

#ML Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
pipeline = Pipeline([
    ("model", LinearRegression())
])
```

#Model Training

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
print(pipeline.predict([[2]]))
```

#Evaluation(metrics)

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print(mean_absolute_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

Here when i removed outliers the score increases for 20 percent

```python
import matplotlib.pyplot as plt
plt.scatter(y_test, y_pred)
```

#Pkl file saved

```python
import pickle
with open("linear_regression_model.pkl", "wb") as file:
    pickle.dump(pipeline, file)
```
